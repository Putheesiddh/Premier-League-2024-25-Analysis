```python
### Q1. Which home teams has the highest expected goal in a single match (XG)?


%%sql
SELECT date, home_team, away_team, score, home_xg
FROM pl_24_25_matches_clean
ORDER BY home_xg DESC
LIMIT 5;
```

```python
### Q2 Every Liverpool home match

%%sql
SELECT date, gameweek, home_xg, home_team, score, away_team, away_xg, venue
FROM pl_24_25_matches_clean
WHERE home_team = 'Liverpool'
ORDER BY date;
```

```python
### Q3 All matches where the away team had a higher xG then the Home Team.

%%sql
SELECT date, home_team, away_team, home_xg, away_xg
FROM pl_24_25_matches_clean
WHERE away_xg > home_xg;
```

```python
### Q4 How many matches of the 380 matches did the away team have a higher xG than the home team?

%%sql
SELECT COUNT(*) AS away_xg_wins
FROM pl_24_25_matches_clean
WHERE away_xg > home_xg;
```

```python
### Q5: Among those 164 matches, which away team has more xG than the home team most often?

%%sql
SELECT away_team, COUNT(*) AS away_xg_wins
FROM pl_24_25_matches_clean
WHERE away_xg > home_xg
GROUP BY away_team
ORDER BY away_xg_wins DESC
LIMIT 5;
```

```python
### Q6 In how many of their 38 matches did Southampton have the lower XG?

%%sql
SELECT COUNT(*) AS lower_xg_matches
FROM pl_24_25_matches_clean
WHERE (home_team = 'Southampton' AND home_xg < away_xg) OR (away_team = 'Southampton' AND away_xg < home_xg);
```

```python
### Q7 In how many of their 38 matches did Liverpool have the lower XG?

%%sql
SELECT COUNT(*) AS lower_xg_matches
FROM pl_24_25_matches_clean
WHERE (home_team = 'Liverpool' AND home_xg < away_xg) OR (away_team = 'Liverpool' AND away_xg < home_xg);
```

```python
### Q8 How many matches did each team have the lower xG in?

%%sql
SELECT team, COUNT(*) AS lower_xg_matches
FROM (SELECT home_team AS team
FROM pl_24_25_matches_clean
WHERE home_xg < away_xg

UNION ALL

SELECT away_team AS team
FROM pl_24_25_matches_clean
WHERE away_xg < home_xg
)
GROUP BY team
ORDER BY lower_xg_matches DESC;
```

```python
### Q9 Splitting the score into home goals and away goals

%%sql
SELECT score,
      INSTR(score, CHAR(8211)) AS dash_position,
      SUBSTR(score, 1, INSTR(score, CHAR(8211)) - 1) AS home_goals,
      SUBSTR(score, INSTR(score, CHAR(8211)) + 1) AS away_goals
FROM pl_24_25_matches_clean
```

```python
%%sql
SELECT COUNT(*) AS bad_scores
FROM pl_24_25_matches_clean
WHERE INSTR(score, CHAR(8211)) = 0;
```

```python
%%sql
CREATE VIEW IF NOT EXISTS matches AS
SELECT *,
    CAST (SUBSTR(score, 1, INSTR(score, CHAR(8211)) - 1) AS INTEGER) AS home_goals,
    CAST (SUBSTR(score, INSTR(score, CHAR(8211)) + 1) AS INTEGER) AS away_goals
FROM pl_24_25_matches_clean;
```

```python
%%sql
SELECT date, home_team, score, away_team, home_goals, away_goals
FROM matches
LIMIT 5;
```

```python
### Q10: How often did the home team actually win?

%%sql
SELECT
      CASE
          WHEN home_goals > away_goals THEN 'Home Win'
          WHEN away_goals > home_goals THEN 'Away Win'
          ELSE 'Draw'
      END AS result, COUNT(*) AS matches
FROM matches
GROUP BY result;
```

```python


%%sql
CREATE VIEW IF NOT EXISTS team_matches AS
SELECT gameweek, date,
       home_team AS team,
       away_team AS opponent,
       'Home'    AS venue,
       home_goals AS goals_for,
       away_goals AS goals_against,
       home_xg AS xg_for,
       away_xg AS xg_against
FROM matches

UNION ALL

SELECT gameweek, date,
       away_team AS team,
       home_team AS opponent,
       'Away'    AS venue,
       away_goals AS goals_for,
       home_goals AS goals_against,
       away_xg AS xg_for,
       home_xg AS xg_against
FROM matches;
```

```python
%%sql
SELECT COUNT(*) AS total_rows FROM team_matches;
```

```python
%%sql
SELECT team, opponent, venue, goals_for, goals_against
FROM team_matches
WHERE team = 'Liverpool' AND opponent = 'Ipswich Town';
```

```python
### Q11: Final standing of the Premier League 24/25

%%sql
SELECT team,
       COUNT(*) AS MP,
       SUM (CASE WHEN  goals_for > goals_against THEN 1 ELSE 0 END) AS W,
       SUM (CASE WHEN goals_for < goals_against THEN 1 ELSE 0 END) AS L,
       SUM (CASE WHEN goals_for = goals_against THEN 1 ELSE 0 END) AS D,
       SUM (goals_for) AS GF,
       SUM (goals_against) AS GA,
       SUM (goals_for) - SUM (goals_against) AS GD,
       SUM (CASE WHEN goals_for > goals_against THEN 3
                 WHEN goals_for = goals_against THEN 1
                 ELSE 0 END) AS PTS


FROM team_matches
GROUP by team
ORDER BY PTS DESC, GD DESC;
```

```python
%%sql
CREATE VIEW IF NOT EXISTS league_table AS
SELECT team,
       COUNT(*) AS MP,
       SUM(CASE WHEN goals_for > goals_against THEN 1 ELSE 0 END) AS W,
       SUM(CASE WHEN goals_for = goals_against THEN 1 ELSE 0 END) AS D,
       SUM(CASE WHEN goals_for < goals_against THEN 1 ELSE 0 END) AS L,
       SUM(goals_for) AS GF,
       SUM(goals_against) AS GA,
       SUM(goals_for) - SUM(goals_against) AS GD,
       SUM(CASE WHEN goals_for > goals_against THEN 3
                WHEN goals_for = goals_against THEN 1
                ELSE 0 END) AS PTS
FROM team_matches
GROUP BY team;
```

```python
%%sql
SELECT * FROM league_table ORDER BY PTS DESC, GD DESC;
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect(f'{PROJECT_DIR}/premier_league.db')
os.makedirs(f'{PROJECT_DIR}/tableau', exist_ok=True)

for name in ['league_table', 'team_matches', 'matches']:
    df = pd.read_sql(f'SELECT * FROM {name}', conn)
    df.to_csv(f'{PROJECT_DIR}/tableau/{name}.csv', index=False)
    print(name, df.shape)

conn.close()
```
